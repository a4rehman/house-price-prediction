````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid", palette="viridis")
````

````python
from src.data.loader import download_raw_data, load_raw_data
from src.data.preprocessing import fill_missing_values, remove_outliers
from src.config import settings

download_raw_data()
df = load_raw_data()
df_clean = remove_outliers(df, target_col=settings.target_column)
df_clean = fill_missing_values(df_clean)
df_clean.shape
````

````python
target = df_clean["SalePrice"]
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
sns.histplot(target, kde=True, ax=axes[0])
axes[0].set_title("SalePrice distribution")
sns.histplot(np.log1p(target), kde=True, ax=axes[1])
axes[1].set_title("log1p(SalePrice) distribution")
plt.tight_layout()
plt.show()
````

````python
missing = df_clean.isna().sum()
missing[missing > 0].sort_values(ascending=False).head(15)
````

````python
numeric = df_clean.select_dtypes(include=[np.number])
corr = numeric.corr()["SalePrice"].drop("SalePrice").abs().sort_values(ascending=False)
corr.head(15)
````

````python
sns.scatterplot(data=df_clean, x="GrLivArea", y="SalePrice", hue="OverallQual", palette="viridis", alpha=0.6)
plt.title("SalePrice vs above-grade living area")
plt.show()
````

````python
sns.boxplot(data=df_clean, x="OverallQual", y="SalePrice")
plt.title("SalePrice by Overall Quality")
plt.ticklabel_format(axis="y", style="sci", scilimits=(0, 0))
plt.show()
````
